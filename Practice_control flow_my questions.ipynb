##APPEND EXAMPLE FROM Google ##
In Python, the term "append" primarily refers to using the list.append() method, which adds a single element to the end of an existing list. This operation modifies the original list in-place. 



Key Characteristics
Adds a Single Element: The append() method takes exactly one argument and adds it as a single item to the end of the list. This argument can be of any data type, including a number, string, or even another list (which results in a nested list).



Modifies In-Place: It directly changes the original list rather than creating a new one.
Returns None: The method does not return the modified list; it returns None. A common mistake is to assign the result of the append operation to a variable (e.g., new_list = old_list.append(item)), which will result in new_list being None.

### 3e. Strings Are Immutable

Calling `.upper()` does **not** modify the original string — it returns a
brand-new string. Every string method that "transforms" actually produces a
new object; the original is always left untouched.
(From Farukhs class work sheet_ control flow)

```python
## Farukhs Example in 3e

name = "alice"
name.upper()          # returns "ALICE" but result is discarded
print(name)           # still "alice"

# To keep the result, assign it:
loud_name = name.upper()
print(loud_name)      # "ALICE"
print(name)           # "alice" -- original still untouched
```
